## Task 2.1 – Inference Pipeline Creation

Now that the model and preprocessing pipeline are registered and staged in MLflow, I will create a reusable inference pipeline. This pipeline will:

1.  Load the staged preprocessing pipeline and model from MLflow.
2.  Define a function to apply the preprocessing and model prediction.
3.  Ensure schema validation and consistent feature ordering for incoming data.

```python
import mlflow.pyfunc

# Load the staged preprocessing pipeline
mlflow_preprocessing_pipeline = mlflow.sklearn.load_model(
    model_uri="models:/TransactionRiskPreprocessingPipeline/Staging"
)
print("Preprocessing pipeline loaded from MLflow (Staging stage).")

# Load the staged model
mlflow_model = mlflow.pyfunc.load_model(
    model_uri="models:/TransactionRiskPredictionModel/Staging"
)
print("Prediction model loaded from MLflow (Staging stage).")
```

```python
def run_inference_pipeline(new_data: pd.DataFrame, expected_columns: list) -> pd.Series:
    """
    Runs the complete inference pipeline on new transaction data.

    Args:
        new_data (pd.DataFrame): New transaction data for inference.
        expected_columns (list): A list of column names representing the expected schema.

    Returns:
        pd.Series: Model predictions (e.g., risk scores or classes).
    """
    # 1. Schema Validation and Feature Ordering
    # Ensure all expected columns are present
    missing_cols = set(expected_columns) - set(new_data.columns)
    if missing_cols:
        raise ValueError(f"Missing columns in new_data: {missing_cols}")

    # Ensure no extra columns are present (optional, but good for strict validation)
    extra_cols = set(new_data.columns) - set(expected_columns)
    if extra_cols:
        print(f"Warning: Extra columns in new_data will be ignored: {extra_cols}")

    # Reorder columns to match the expected schema
    inference_data_ordered = new_data[expected_columns].copy()

    # 2. Apply Preprocessing Pipeline
    # The mlflow_preprocessing_pipeline is the scikit-learn pipeline object
    preprocessed_data = mlflow_preprocessing_pipeline.transform(inference_data_ordered)
    print("Data preprocessed successfully.")

    # 3. Make Predictions
    # The mlflow_model is an MLflow PyFuncModel, which has a predict method
    predictions = mlflow_model.predict(preprocessed_data)
    print("Predictions generated successfully.")

    return pd.Series(predictions)

print("Inference pipeline function defined.")
```

```python
# Demonstrate the inference pipeline with new dummy data
# Use the same 'dummy_cols' that were defined during validation to ensure schema consistency
# new_dummy_data = pd.DataFrame(0, index=[0, 1, 2], columns=dummy_cols) # Example with all zeros

# For better demonstration, let's create slightly different dummy data
new_dummy_data = pd.DataFrame({
    col: [1 if i % 2 == 0 else 0 for i in range(3)] for i, col in enumerate(dummy_cols)
})
new_dummy_data['TransactionAmt'] = [100.0, 250.5, 50.0]

print("New dummy data for inference:")
display(new_dummy_data.head())

try:
    inference_results = run_inference_pipeline(new_dummy_data, dummy_cols)
    print("\nInference Results:")
    display(inference_results)
except ValueError as e:
    print(f"Error during inference: {e}")
```

## Task 2.2 – REST API Deployment

Now, let's expose the inference pipeline through a REST API using FastAPI. The API will accept transaction-level input data, run it through our `run_inference_pipeline`, and return a risk score (prediction) and a risk flag (`IsFraud`).

```python
# Install FastAPI and uvicorn
!pip install fastapi uvicorn nest_asyncio
```

```python
from fastapi import FastAPI
from pydantic import BaseModel
import nest_asyncio
import uvicorn
import pandas as pd

# Apply nest_asyncio to allow uvicorn to run in a Jupyter environment
nest_asyncio.apply()

app = FastAPI(title="Transaction Risk Prediction API")

print("FastAPI app initialized.")
```

```python
from pydantic import BaseModel, Field, create_model

# Dynamically create a Pydantic model based on dummy_cols
# This ensures the API input schema matches the expected features.

# Create a dictionary of fields for create_model
# Each field will be annotated as float with a default value of 0.0
model_fields = {col: (float, Field(default=0.0)) for col in dummy_cols}

TransactionData = create_model('TransactionData', **model_fields)

print("Pydantic model 'TransactionData' created dynamically based on dummy_cols.")
```

```python
from datetime import datetime
from typing import List

# Define a Pydantic model for the log entry
class PredictionLog(BaseModel):
    timestamp: datetime
    transaction_data: TransactionData # Using the dynamically created TransactionData model
    risk_score: float
    is_fraud: bool
    message: str

# Global list to store prediction logs (for demonstration purposes)
prediction_logs: List[PredictionLog] = []

print("PredictionLog model and global prediction_logs list initialized.")
```

```python
@app.post("/predict")
async def predict_risk(transaction_data: TransactionData):
    """
    Accepts transaction data, predicts risk, and returns a risk score .
    """
    current_timestamp = datetime.now() # Capture timestamp at the start of prediction

    # Convert incoming Pydantic model data to a Pandas DataFrame
    input_df = pd.DataFrame([transaction_data.model_dump()]) # Changed .dict() to .model_dump()

    # Ensure the DataFrame columns are in the correct order as expected by the pipeline
    # The `run_inference_pipeline` function already handles column ordering and validation
    try:
        risk_predictions = run_inference_pipeline(input_df, expected_columns=dummy_cols)
        # Assuming risk_predictions are 0 or 1 (binary classification)
        is_fraud_flag = bool(risk_predictions.iloc[0]) # Convert prediction to boolean flag
        risk_score = float(risk_predictions.iloc[0]) # Assuming a binary prediction (0 or 1)

        # Create a log entry
        log_entry = PredictionLog(
            timestamp=current_timestamp,
            transaction_data=transaction_data, # Store the original Pydantic model data
            risk_score=risk_score,
            is_fraud=is_fraud_flag,
            message="Prediction successful"
        )
        prediction_logs.append(log_entry)

        return {"risk_score": risk_score, "IsFraud": is_fraud_flag}
    except ValueError as e:
        error_message = f"Input data schema mismatch: {e}"
        log_entry = PredictionLog(
            timestamp=current_timestamp,
            transaction_data=transaction_data,
            risk_score=0.0, # Default or placeholder for failed predictions
            is_fraud=False, # Default or placeholder for failed predictions
            message=error_message
        )
        prediction_logs.append(log_entry)
        return {"error": str(e), "message": "Input data schema mismatch."}
    except Exception as e:
        error_message = f"An unexpected error occurred during prediction: {e}"
        log_entry = PredictionLog(
            timestamp=current_timestamp,
            transaction_data=transaction_data,
            risk_score=0.0, # Default or placeholder for failed predictions
            is_fraud=False, # Default or placeholder for failed predictions
            message=error_message
        )
        prediction_logs.append(log_entry)
        return {"error": str(e), "message": "An unexpected error occurred during prediction."}

print("'/predict' API endpoint defined and updated for logging.")
```

```python
@app.get("/logs", response_model=List[PredictionLog])
async def get_prediction_logs():
    """
    Retrieves all stored prediction logs.
    """
    return prediction_logs

print("'/logs' API endpoint defined.")
```

```python
data_drift_report = Report(metrics=[DataDriftPreset()])
data_drift_report.run(reference_data=reference_data, current_data=current_data, column_mapping=None)

# The report can be displayed in the notebook or saved as HTML
data_drift_report.show()
```

This report helps to quickly identify features that have significantly changed their distribution, which could impact model performance.
