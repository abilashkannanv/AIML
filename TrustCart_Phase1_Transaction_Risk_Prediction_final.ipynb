## Task 1.2 – Data Joining Strategy

```python
# Identify common keys
common_keys = list(set(transactions_df.columns) & set(user_identity_df.columns))
print(f"Common key(s) between transactions_df and user_identity_df: {common_keys}")

# Decide join type and justify choice
# Since user_identity.csv contains additional identity information for a subset of transactions,
# and we want to retain all transaction data while enriching it with identity details where available,
# a left join on 'TransactionID' is appropriate.
# This will keep all records from transactions_df and add matching identity information from user_identity_df.
# Non-matching identity records will have NaN values for identity-specific columns.

merged_df = pd.merge(transactions_df, user_identity_df, on='TransactionID', how='left')

print('\nMerged DataFrame Head:')
display(merged_df.head())

print('\nMerged DataFrame Info:')
display(merged_df.info())

print('\nMerged DataFrame Shape:')
print(merged_df.shape)
```

## Task 2.2 – Data Cleaning

```python
# Identify columns with very high missing percentages to drop
missing_percentages = (merged_df.isnull().sum() / len(merged_df)) * 100
columns_to_drop_high_missing = missing_percentages[missing_percentages > 90].index.tolist()

print(f"Columns to drop due to more than 90% missing values: {len(columns_to_drop_high_missing)} columns")
print(columns_to_drop_high_missing)

# Drop these columns
merged_df_cleaned = merged_df.drop(columns=columns_to_drop_high_missing)
print(f"Shape after dropping high-missing columns: {merged_df_cleaned.shape}")

# Separate numerical and categorical columns for imputation
# Exclude 'TransactionID' and 'isFraud' (target) from imputation for now if they are not missing
numerical_cols = merged_df_cleaned.select_dtypes(include=['number']).columns.tolist()
categorical_cols = merged_df_cleaned.select_dtypes(include=['object']).columns.tolist()

# Remove 'TransactionID' and 'isFraud' if they are in numerical_cols and should not be imputed this way
if 'TransactionID' in numerical_cols: numerical_cols.remove('TransactionID')
# Assuming 'isFraud' is our target 'risk_flag' and should not be imputed
if 'isFraud' in numerical_cols: numerical_cols.remove('isFraud')

print(f"\nNumber of numerical columns for imputation: {len(numerical_cols)}")
print(f"Number of categorical columns for imputation: {len(categorical_cols)}")

# Impute numerical columns with median
for col in numerical_cols:
    if merged_df_cleaned[col].isnull().any():
        median_val = merged_df_cleaned[col].median()
        merged_df_cleaned[col].fillna(median_val, inplace=True)

# Impute categorical columns with mode
for col in categorical_cols:
    if merged_df_cleaned[col].isnull().any():
        mode_val = merged_df_cleaned[col].mode()[0] # .mode() can return multiple values, take the first
        merged_df_cleaned[col].fillna(mode_val, inplace=True)

print('\nMissing values after cleaning:')
print(merged_df_cleaned.isnull().sum().sum())

print('\nCleaned DataFrame Info (first few rows):')
display(merged_df_cleaned.head())

print('\nCleaned DataFrame Shape:')
print(merged_df_cleaned.shape)

# Prevent target leakage:
# The target variable 'isFraud' (risk_flag) should not be used to create or impute features.
# Our current steps already avoid this by excluding 'isFraud' from imputation steps for features.
# Also, no future features will be derived directly from 'isFraud' for training.
```

## Task 6.1 – Metric Selection

```python
print('### Business-Appropriate Metrics for Fraud Detection ###')

print('\nGiven the highly imbalanced nature of fraud detection datasets (where fraudulent transactions are a small minority), traditional metrics like accuracy can be misleading. Therefore, we will focus on the following business-appropriate metrics:')

print('\n1.  **ROC AUC (Receiver Operating Characteristic Area Under the Curve):**')
print('    *   **Description:** Measures the ability of the model to distinguish between positive (fraud) and negative (non-fraud) classes across all possible classification thresholds. It plots the True Positive Rate (TPR) against the False Positive Rate (FPR).')
print('    *   **Justification:** A high ROC AUC indicates that the model is good at separating the two classes, regardless of the chosen threshold. It is robust to class imbalance.')

print('\n2.  **Precision-Recall AUC (Area Under the Precision-Recall Curve):**')
print('    *   **Description:** Plots precision (proportion of correctly identified positive predictions that were actually positive) against recall (proportion of actual positive cases that were correctly identified).')
print('    *   **Justification:** For highly imbalanced datasets, PR AUC is often a more informative metric than ROC AUC, especially when the positive class is rare and false positives are costly. It focuses on the performance of the model on the minority class. A model that achieves high precision and high recall on the minority class is desirable.')

print('\n3.  **Recall (Sensitivity or True Positive Rate) for the Minority Class (Fraud):**')
print('    *   **Description:** The proportion of actual fraudulent transactions that the model correctly identified as fraudulent.')
print('    *   **Justification:** From a business perspective, **missing actual fraud (false negatives) can be very costly**. Therefore, maximizing recall for the fraud class is often a primary objective, even if it means tolerating a certain level of false positives (lower precision).')

print('\n4.  **Precision for the Minority Class (Fraud):**')
print('    *   **Description:** The proportion of predicted fraudulent transactions that were actually fraudulent.')
print("    *   **Justification:** While recall is critical, precision is also important. **A very low precision means many legitimate transactions are incorrectly flagged as fraud**, leading to high operational costs (manual review, customer inconvenience, false alarms). A balance between precision and recall is often sought, depending on the business's risk tolerance and operational capacity.")

print('\n5.  **F1-Score for the Minority Class (Fraud):**')
print('    *   **Description:** The harmonic mean of precision and recall. It provides a single score that balances both concerns.')
print('    *   **Justification:** Useful when we need a balance between precision and recall for the minority class. A higher F1-score indicates better performance in scenarios where both false positives and false negatives are important.')

print('\nWe will primarily use **ROC AUC** and **Precision-Recall AUC** for overall model comparison, and then delve into **Recall** and **Precision** to understand the specific trade-offs and business implications of each model.')
```

```python
import pandas as pd # Ensure pandas is imported
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, precision_recall_curve, auc
from imblearn.over_sampling import SMOTE
from collections import Counter
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Re-initialization block for X_transformed and y, X_train, X_test, y_train, y_test, X_train_res, y_train_res
# This is crucial for robustness if the kernel state is not maintained.
# This block is a copy of the one used in Task 5.2 and 6.2 to ensure data availability.

if 'X_transformed' not in locals() or 'y' not in locals():
    print("Re-running feature processing steps and data split for model explainability...")
    transactions_df = pd.read_csv('/content/sample_data/transaction.csv')
    user_identity_df = pd.read_csv('/content/sample_data/identity.csv')
    merged_df = pd.merge(transactions_df, user_identity_df, on='TransactionID', how='left')

    missing_percentages = (merged_df.isnull().sum() / len(merged_df)) * 100
    columns_to_drop_high_missing = missing_percentages[missing_percentages > 90].index.tolist()
    merged_df_cleaned = merged_df.drop(columns=columns_to_drop_high_missing)

    numerical_cols_temp = merged_df_cleaned.select_dtypes(include=['number']).columns.tolist()
    categorical_cols_temp = merged_df_cleaned.select_dtypes(include=['object']).columns.tolist()

    if 'TransactionID' in numerical_cols_temp: numerical_cols_temp.remove('TransactionID')
    if 'isFraud' in numerical_cols_temp: numerical_cols_temp.remove('isFraud')

    for col in numerical_cols_temp:
        if merged_df_cleaned[col].isnull().any():
            median_val = merged_df_cleaned[col].median()
            merged_df_cleaned[col].fillna(median_val, inplace=True)

    for col in categorical_cols_temp:
        if merged_df_cleaned[col].isnull().any():
            mode_val = merged_df_cleaned[col].mode()[0]
            merged_df_cleaned[col].fillna(mode_val, inplace=True)

    # Derived Features (as in Task 4.2)
    X = merged_df_cleaned.drop(columns=['isFraud', 'TransactionID'])
    y = merged_df_cleaned['isFraud']

    SECONDS_IN_DAY = 24 * 60 * 60
    X['seconds_in_day'] = X['TransactionDT'] % SECONDS_IN_DAY
    X['hour_of_day'] = X['seconds_in_day'] // 3600
    X['day_of_week'] = (X['TransactionDT'] // SECONDS_IN_DAY) % 7
    X['TransactionAmt_to_Card1_mean'] = X['TransactionAmt'] / X.groupby(['card1'])['TransactionAmt'].transform('mean')
    X['TransactionAmt_to_Card1_std'] = X['TransactionAmt'] / X.groupby(['card1'])['TransactionAmt'].transform('std')
    X['TransactionAmt_to_Card1_std'].fillna(0, inplace=True)
    X['card1_transactions_per_hour_day'] = X.groupby(['card1', 'day_of_week', 'hour_of_day'])['TransactionAmt'].transform('count')

    numerical_features_final = X.select_dtypes(include=['number']).columns.tolist()
    categorical_features_final = X.select_dtypes(include=['object']).columns.tolist()

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), numerical_features_final),
            ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features_final)
        ],
        remainder='passthrough'
    )
    preprocessing_pipeline = Pipeline(steps=[('preprocessor', preprocessor)])
    X_transformed = preprocessing_pipeline.fit_transform(X)
    print("Feature processing complete.")

    X_train, X_test, y_train, y_test = train_test_split(X_transformed, y, test_size=0.3, random_state=42, stratify=y)

    smote = SMOTE(random_state=42)
    X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
    print("Data split and SMOTE resampling complete.")

    # Train the models again to ensure they are available
    # Logistic Regression
    model_lr = LogisticRegression(solver='liblinear', random_state=42, class_weight='balanced', n_jobs=-1)
    model_lr.fit(X_train_res, y_train_res)

    # Random Forest
    rf_model_explain = RandomForestClassifier(random_state=42, n_jobs=-1, class_weight='balanced', n_estimators=100, max_depth=10, min_samples_leaf=5) # Using simplified best params
    rf_model_explain.fit(X_train_res, y_train_res)

    # XGBoost
    xgb_model_explain = XGBClassifier(random_state=42, n_jobs=-1, use_label_encoder=False, eval_metric='logloss', n_estimators=100, max_depth=3, learning_rate=0.1) # Using simplified best params
    xgb_model_explain.fit(X_train_res, y_train_res)
    print("Models re-trained for explainability.")


print('\n### Analyzing Feature Importance from the Best Model (e.g., XGBoost) ###')

# Get feature names after one-hot encoding for the preprocessor pipeline
# This assumes 'preprocessor' and 'preprocessing_pipeline' are available from the robust block
feature_names = []

# Get numerical feature names
numerical_feature_names = preprocessing_pipeline.named_steps['preprocessor'].named_transformers_['num'].get_feature_names_out(numerical_features_final)
feature_names.extend(numerical_feature_names)

# Get categorical feature names after one-hot encoding
categorical_transformer = preprocessing_pipeline.named_steps['preprocessor'].named_transformers_['cat']
categorical_feature_names = categorical_transformer.get_feature_names_out(categorical_features_final)
feature_names.extend(categorical_feature_names)

# Assign to a variable that will hold the best model for feature importance
# For this example, let's assume XGBoost was the best performing one in terms of PR AUC.
# In a real scenario, you would dynamically select the best_model based on Task 6.2 results.
best_model = xgb_model_explain # Using the re-trained XGBoost model

if hasattr(best_model, 'feature_importances_'):
    importances = best_model.feature_importances_
elif hasattr(best_model, 'coef_'):
    # For linear models like Logistic Regression, coefficients can be used as importance
    # Need to handle multi-class vs binary and potentially absolute values
    if best_model.coef_.shape[0] == 1:
        importances = np.abs(best_model.coef_[0])
    else: # Multi-class, may need to average or take specific class coef
        importances = np.mean(np.abs(best_model.coef_), axis=0)
else:
    print("Selected model does not have 'feature_importances_' or 'coef_'. Cannot show feature importance.")
    importances = None

if importances is not None:
    feature_importance_df = pd.DataFrame({
        'Feature': feature_names,
        'Importance': importances
    })

    # Sort by importance and display top N features
    feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

    print('\nTop 20 Features by Importance:')
    display(feature_importance_df.head(20))

    # Visualize feature importance
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Importance', y='Feature', data=feature_importance_df.head(20))
    plt.title('Top 20 Feature Importances')
    plt.xlabel('Importance')
    plt.ylabel('Feature')
    plt.tight_layout()
    plt.show()

    print('\n### Discussion of Feature Importance ###')
    print('Feature importance analysis helps in understanding which variables contribute most to the model\'s predictions. This can provide valuable business insights, such as identifying critical transaction attributes or user behaviors associated with fraud. For example, high importance of `TransactionAmt_to_Card1_mean` might indicate that deviations from typical spending patterns are strong indicators of fraud.')
    print('This information can also be used for feature selection, model simplification, or to guide further data collection efforts. It helps stakeholders trust and interpret the model\'s decisions.')
```
