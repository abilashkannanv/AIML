```python
# execute to load the dataframes correctly
transactions_df = pd.read_csv('sample_data/transaction.csv')
user_identity_df = pd.read_csv('sample_data/identity.csv')

print('Transactions DataFrame:')
display(transactions_df.head())
print('\nTransactions DataFrame Info:')
transactions_df.info()

print('\nUser Identity DataFrame:')
display(user_identity_df.head())
print('\nUser Identity DataFrame Info:')
user_identity_df.info()

target_variable = 'risk_flag'

if target_variable in transactions_df.columns:
    print(f"\nTarget variable '{target_variable}' found in transactions_df.")
    print(f"Number of records in transactions_df: {len(transactions_df)}")
    print(f"Number of features in transactions_df (including target): {len(transactions_df.columns)}")
elif target_variable in user_identity_df.columns:
    print(f"\nTarget variable '{target_variable}' found in user_identity_df.")
    print(f"Number of records in user_identity_df: {len(user_identity_df)}")
    print(f"Number of features in user_identity_df (including target): {len(user_identity_df.columns)}")
else:
    print(f"\nTarget variable '{target_variable}' not found in either DataFrame.")
```

### Task 1.2: Identify common keys and decide join type

```python
print('Common columns between transactions_df and user_identity_df:')
common_cols = set(transactions_df.columns) & set(user_identity_df.columns)
print(common_cols)

# Assuming 'User ID' or a similar column is the common key
# Let's inspect unique values in potential key columns if they exist

# Example of potential common keys (adjust based on actual common_cols)
if 'User ID' in common_cols:
    print(f"\nNumber of unique 'User ID' in transactions_df: {transactions_df['User ID'].nunique()}")
    print(f"Number of unique 'User ID' in user_identity_df: {user_identity_df['User ID'].nunique()}")

# Justify join type (assuming 'User ID' is the key)
# If we want to keep all transactions and add user identity where available, a left join is suitable.
# If we only want to keep records where both transaction and identity data exist, an inner join is suitable.
# If we want all user identities and corresponding transactions, a right join might be considered.
# If we want all records from both and fill missing with NaN, a full outer join.

# For this task, a left join from transactions_df to user_identity_df seems appropriate
# if the goal is to enrich transaction data with user identity information.
# However, without specific instructions on which table to prioritize or what analysis to perform, a general approach is to understand the overlap.

print("\nDecision for Join Type and Justification:")
print("Based on the likely goal of enriching transaction data with user details, a **left join** is typically appropriate. This ensures all transactions are retained, and user identity information is added where a match exists. If a 'User ID' from transactions_df does not have a corresponding entry in user_identity_df, the identity columns will have NaN values. If the goal was to analyze only users for whom we have both transaction and identity data, an **inner join** would be used.")
print("Let's assume 'User ID' is the common key for joining and proceed with a left join to ensure all transaction data is kept.")

# Perform a sample left join to demonstrate (optional, can be done in next step if user requests merge)
# merged_df = pd.merge(transactions_df, user_identity_df, on='User ID', how='left')
# display(merged_df.head())
# print(merged_df.info())
```

```python
# Perform the left join
merged_df = pd.merge(transactions_df, user_identity_df, on='TransactionID', how='left')

print('Merged DataFrame Head:')
display(merged_df.head())

print('\nMerged DataFrame Info:')
merged_df.info()

print(f"\nNumber of records in merged_df: {len(merged_df)}")
print(f"Number of features in merged_df: {len(merged_df.columns)}")
```

## Task 2.1 – Missing Value Analysis

```python
# Calculate missing values and their percentages
missing_values = merged_df.isnull().sum()
missing_percentages = (merged_df.isnull().sum() / len(merged_df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentages
})

# Filter to show only columns with missing values and sort them
missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print("Columns with Missing Values and their Percentages:")
display(missing_info.head(20)) # Display top 20 columns with most missing values

# Categorize missingness (general observation)
print("\nCategorizing Missingness:")
high_missing = missing_info[missing_info['Missing Percentage'] > 70]
medium_missing = missing_info[(missing_info['Missing Percentage'] <= 70) & (missing_info['Missing Percentage'] > 30)]
low_missing = missing_info[missing_info['Missing Percentage'] <= 30]

if not high_missing.empty:
    print("\nHighly Missing Columns ( > 70% missing):")
    display(high_missing.head())
if not medium_missing.empty:
    print("\nModerately Missing Columns (30% - 70% missing):")
    display(medium_missing.head())
if not low_missing.empty:
    print("\nLow Missing Columns (<= 30% missing):")
    display(low_missing.head())
```

## Task 2.2 – Data Cleaning

### Handling Missing Values and Dropping/Retaining Columns

Given the high percentage of missing values in many columns, especially those from the `identity_df` that had no match in `transactions_df` (resulting in ~75-99% missing data), it's advisable to drop columns that have an excessively high percentage of missing values as they are unlikely to contribute meaningfully to the model and can introduce noise. We will set a threshold for dropping columns.

Let's start by dropping columns with more than, say, 80% missing values. We will then decide on a strategy for the remaining missing values.

```python
# Define a threshold for dropping columns (e.g., 80% missing)
missing_threshold = 80

# Get columns to drop based on the threshold
columns_to_drop = missing_info[missing_info['Missing Percentage'] > missing_threshold].index.tolist()

print(f"Number of columns to drop (>{missing_threshold}% missing): {len(columns_to_drop)}")
print("Columns to be dropped:")
for col in columns_to_drop:
    print(f"- {col} ({missing_info.loc[col]['Missing Percentage']:.2f}% missing)")

# Drop the identified columns from merged_df
df_cleaned = merged_df.drop(columns=columns_to_drop)

print(f"\nShape of merged_df before dropping: {merged_df.shape}")
print(f"Shape of df_cleaned after dropping highly missing columns: {df_cleaned.shape}")

# Update missing_info for the cleaned DataFrame
missing_values_cleaned = df_cleaned.isnull().sum()
missing_percentages_cleaned = (df_cleaned.isnull().sum() / len(df_cleaned)) * 100

missing_info_cleaned = pd.DataFrame({
    'Missing Count': missing_values_cleaned,
    'Missing Percentage': missing_percentages_cleaned
})

missing_info_cleaned = missing_info_cleaned[missing_info_cleaned['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print("\nMissing values in df_cleaned after dropping high-missing columns:")
display(missing_info_cleaned.head(20))

# For the remaining missing values (moderately and low missing), we can consider imputation.
# Let's inspect the data types of columns with remaining missing values to choose an appropriate imputation strategy.

print("\nData types of columns with remaining missing values:")
display(df_cleaned[missing_info_cleaned.index].dtypes)
```

### Handle Remaining Missing Values (Imputation)

```python
# Separate numerical and categorical columns for imputation

numerical_cols_with_missing = missing_info_cleaned[df_cleaned[missing_info_cleaned.index].dtypes != 'object'].index.tolist()
categorical_cols_with_missing = missing_info_cleaned[df_cleaned[missing_info_cleaned.index].dtypes == 'object'].index.tolist()

print(f"Number of numerical columns with remaining missing values: {len(numerical_cols_with_missing)}")
print(f"Number of categorical columns with remaining missing values: {len(categorical_cols_with_missing)}")

# Impute numerical columns with the median
print("\nImputing numerical columns with median...")
for col in numerical_cols_with_missing:
    if col in df_cleaned.columns:
        median_val = df_cleaned[col].median()
        df_cleaned[col].fillna(median_val, inplace=True)
        # print(f"  Imputed '{col}' with median: {median_val}")

# Impute categorical columns with the mode
print("\nImputing categorical columns with mode...")
for col in categorical_cols_with_missing:
    if col in df_cleaned.columns:
        mode_val = df_cleaned[col].mode()[0] # .mode() can return multiple if frequencies are tied
        df_cleaned[col].fillna(mode_val, inplace=True)
        # print(f"  Imputed '{col}' with mode: {mode_val}")

print("\nMissing value imputation complete. Verifying...")

# Verify no more missing values in df_cleaned
missing_after_imputation = df_cleaned.isnull().sum()
missing_after_imputation = missing_after_imputation[missing_after_imputation > 0]

if missing_after_imputation.empty:
    print("All missing values have been handled.")
else:
    print("Remaining missing values (should be none if all categories handled):")
    display(missing_after_imputation)

print(f"\nFinal shape of df_cleaned: {df_cleaned.shape}")
display(df_cleaned.head())
```
